```python
import pandas as pd
```

```python
# Load the training log
results = pd.read_csv('runs/detect/bumble_bee-wasps-bee-dataset.yolov8x/100epochs_imgsz640_patience30/results.csv')

# Strip spaces
results.columns = results.columns.str.strip()

# Calculate fitness
results["fitness"] = results["metrics/mAP50(B)"] * 0.1 + results["metrics/mAP50-95(B)"] * 0.5 + results["metrics/recall(B)"] * 0.2 + results["metrics/precision(B)"] * 0.2

# Find the epoch with the highest fitness
best_epoch = results['fitness'].idxmax() + 1

print(f"This model had it's highest fitness at epoch: {best_epoch}, with a fitness of: {results['fitness'].max()}")
```
